This notebook demonstrates how to **link cloud storage (Azure Data Lake) to Databricks** using Unity Catalog.

**What we do here:**
1. Verify we can see files in Azure storage
2. Create an external location (a secure connection between Databricks and Azure)

This is a practice/demo notebook using the `demo` container.

```python
%fs ls 'abfss://demo@databricksprojectsextdl1.dfs.core.windows.net/'
```

### Create an External Location
An external location registers a cloud storage path in Unity Catalog. It uses a **storage credential** (a pre-configured authentication method) to securely access Azure without exposing any keys or secrets in your code.

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_projects_ext_dl1_demo
    URL 'abfss://demo@databricksprojectsextdl1.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-course-sc`)
    COMMENT 'External location for databricks projects';
```
