```python
dbutils.widgets.text('p_batch_id','')
v_batch_id= dbutils.widgets.get('p_batch_id')
```

```python
from pyspark.sql.functions import *
from pyspark.sql import functions as f
```

```python
bronze_table = f'{catalog_name}.{bronze_schema}.results'
silver_table = f'{catalog_name}.{silver_schema}.results'
```

#### Read Bronze

```python
# Read Bronze table
results_df = spark.read.table(bronze_table).filter(col('batch_id') == v_batch_id)
```

#### Drop Columns
- Remove `url` column — not needed for analysis

```python
# Drop unnecessary columns
results_drop_df = results_df.drop('url')
```

#### Rename Columns
- Convert camelCase to snake_case for consistency

```python
# Rename columns to snake_case
results_renamed_df = (results_drop_df
    .withColumnsRenamed({'constructorId': 'constructor_id',
                         'driverId': 'driver_id',
                         'raceName': 'race_name',
                         'date': 'race_date',
                         'grid' : 'grid_position',
                         'laps':'Completed_laps',
                         'number' : 'car_number',
                         'position':'final_position',
                         'positionText':'final_position_text'})
)
display(results_renamed_df)
```

#### Remove Nulls
- Drop rows where key columns (`season`, `round`, `constructor_id`, `driver_id`) are null

```python
# Remove rows with null key values
results_valid_df = (
    results_renamed_df
    .filter(
        col('season').isNotNull() &
        col('round').isNotNull() &
        col('constructor_id').isNotNull() &
        col('driver_id').isNotNull()
    )
)
display(results_valid_df)
```

#### Remove Duplicates
- Keep one row per driver per race using `dropDuplicates()`

```python
# Remove duplicates
results_distinct_df = results_valid_df.dropDuplicates(['driver_id','season','constructor_id','round'])
display(results_distinct_df)
```

#### Title Case
- Apply `initcap()` to `race_name` so it looks clean

```python
# Apply title case to race_name
results_final_df = results_distinct_df.withColumn('race_name', initcap(col('race_name')))
```

#### Write to Silver
- If the silver table doesn't exist yet, it creates it from scratch
- If it already exists, it merges new/updated rows using `write_to_silver()` (insert new, update changed)

```python
write_to_silver(
    input_df=results_final_df,
    target_table=silver_table,
    merge_condition= """
    s.season = t.season 
    AND s.round = t.round 
    AND s.constructor_id = t.constructor_id 
    AND s.driver_id = t.driver_id""",
    
    columns_to_update=[
    "race_date",
"race_name",
"round",
"season",
"constructor_id",
"driver_id",
"grid_position",
"Completed_laps",
"car_number",
"points",
"final_position",
"final_position_text",
"status",
"ingestion_timestamp",
"source_file",
"batch_id"
    ]
)
```

#### Verify
- Read back the silver table to confirm data was saved

```python
# Verify
spark.read.table(silver_table).display()
```
