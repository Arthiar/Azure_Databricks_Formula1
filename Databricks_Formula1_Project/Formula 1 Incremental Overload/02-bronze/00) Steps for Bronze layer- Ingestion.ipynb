**Bronze Layer - Formula 1 Incremental**

Loads raw files (CSV/JSON) from the batch landing folder into `formula1_incr.bronze` Delta tables, partitioned by `batch_id` for incremental processing.

---

**Steps in each notebook:**
1. Load config (`%run ../00-common/01.environment-config`)
2. Get `batch_id` from widget parameter
3. Set source path: `{loding_folder_path}/{batch_id}/filename`
4. Define schema
5. Read file
6. Add metadata (`ingestion_timestamp`, `source_file`)
7. Add `batch_id` column and write to Delta (partitioned, overwrite per batch)

---

| Notebook | File | Format | Schema Type |
| --- | --- | --- | --- |
| 01 | circuits.csv | CSV | StructType |
| 02 | races.csv | CSV | StructType |
