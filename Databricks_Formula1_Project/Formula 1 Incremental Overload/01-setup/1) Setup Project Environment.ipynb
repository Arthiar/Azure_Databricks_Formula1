**Formula 1 Incremental - Environment Setup**

Catalog: `formula1_incr` | Container: `formula1-incremental` | Storage: Azure Data Lake

**Verify storage access**

```python
%fs ls 'abfss://formula1-incremental@databricksprojectsextdl1.dfs.core.windows.net/landing'
```

**Create external location**

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_projects_ext_dl1_formula1_incr
    URL 'abfss://formula1-incremental@databricksprojectsextdl1.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-course-sc`)
    COMMENT 'External location for databricks projects';
```

**Create catalog**

```python
%sql
CREATE CATALOG IF NOT EXISTS  formula1_incr
   MANAGED LOCATION 'abfss://formula1-incremental@databricksprojectsextdl1.dfs.core.windows.net/' 
   COMMENT 'This is the main catalog for our Formula 1 incremental project';
```

**View existing catalogs**

```python
%sql
SHOW CATALOGS
```

**Create schemas**

```python
%sql
CREATE  SCHEMA  IF NOT EXISTS formula1_incr.landing;
CREATE  SCHEMA  IF NOT EXISTS formula1_incr.bronze
     MANAGED LOCATION 'abfss://formula1-incremental@databricksprojectsextdl1.dfs.core.windows.net/bronze';
CREATE  SCHEMA  IF NOT EXISTS formula1_incr.silver
     MANAGED LOCATION 'abfss://formula1-incremental@databricksprojectsextdl1.dfs.core.windows.net/silver';
CREATE  SCHEMA  IF NOT EXISTS formula1_incr.gold
     MANAGED LOCATION 'abfss://formula1-incremental@databricksprojectsextdl1.dfs.core.windows.net/gold'
```

**Verify schemas**

```python
%sql
use catalog formula1_incr;
show schemas;
```

**Create volume**

```python
%sql
CREATE EXTERNAL VOLUME IF NOT EXISTS formula1_incr.landing.files
LOCATION 'abfss://formula1-incremental@databricksprojectsextdl1.dfs.core.windows.net/landing';
```
