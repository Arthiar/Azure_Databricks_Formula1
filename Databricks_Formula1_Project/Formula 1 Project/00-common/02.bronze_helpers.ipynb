## Bronze Layer - Helper Functions

This notebook contains **reusable functions** shared across all Bronze ingestion notebooks (circuits, races, drivers, etc.).

**Why a shared helper file?** Every Bronze table needs the same metadata columns added. Instead of copying the same code into every notebook, we write it once here and call `%run` to import it.

#### What the function does:
`add_ingerstion_metadata(df)` takes any DataFrame and adds two columns:
- **`ingestion_timestamp`** - the exact time the data was loaded
- **`source_file`** - the file path the row came from

This helps us track *when* and *where* each record was ingested.

```python
# Helpers function to add the file metadata and ingestion timestamp

from pyspark.sql.functions import *

def add_ingestion_metadata(df):
    return(df
        .withColumn('ingestion_timestamp', current_timestamp())
        .withColumn('source_file', col('_metadata.file_path'))
    )
```
